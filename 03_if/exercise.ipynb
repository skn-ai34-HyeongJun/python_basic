```python
print("=== OhGiraffers vending machine ===")
print("tiger: 우리는 1인 24시간 교대제로 간다")
print("monkey: 24시간 문의 환영, 다른 말로 하면?")
print("squirrel: 저도 하루에 너다섯 시간 자는데요?")
print("capybara: 살아남는 사람이 강한 사람입니다.")
print("====================================")
print("만나고 싶은 선생님을 입력해 주세요")
input_teacher = input()
print("====================================")

sleep_time = 0
match input_teacher:
    case "tiger":
        print("호랑이 선생님의 메타버스에 오신 것을 환영합니다.")
        sleep_time = 3
    case "monkey":
        print("원숭이 선생님의 밀림에 오신 것을 환영합니다.")
        sleep_time = 4
    case "squirrel":
        print("다람쥐 선생님의 매력의 늪에 오신 것을 환영합니다.")
        sleep_time = 5
    case "capybara" :  
        print("카피바라 선생님의 교실에 오신 것을 환영합니다.")
        sleep_time = 6
    case _:
        print("공부할 마음이 없으신 건 아니죠?")

print("매일 " + str(sleep_time) + "시간의 수면이 예상됩니다.")
# print("매일", sleep_time, "시간의 수면이 예상됩니다.")
```

```python
# 1~10까지 제곱값을 담은 리스트 만들기
lst = []
for n in range(1,11):
    lst.append(n ** 2)
print(lst)

# 리스트 내포 (List Comprehension)
lst = [n*n for n in range(1,11)]
print(lst)

lst = [n*n for n in range(1,11) if n % 2 == 0]
print(lst)
```

```python
# 실습문제 : Life is too short, you need python에서 aeiou제거하기

# 방법1
s = 'Life is too short, you need python'
print([x for x in s if(x not in ['a','e','i','o','u'])])

# 방법2
vowels = 'aeiou'
print(''.join([a for a in s if a not in vowels]))
```
