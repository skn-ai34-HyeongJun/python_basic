문제 1: 클래스 속성, 인스턴스 속성 테스트  
카피바라는 이직 준비를 위해 자기소개서를 작성해야 합니다. Squirrel이라는 클래스를 만들어 카피바라의 기본 정보를 설정하세요.  
클래스 속성 species를 설정하여 "카피바라"로 초기화하세요.  
생성자를 통해 인스턴스 속성 name과 current_job을 설정하세요.  
메서드 introduce를 만들어 이름과 현재 직업을 소개하는 문자열을 반환하세요.  
다른 인스턴스를 생성해도 species는 공유되지만, name과 current_job은 서로 독립적임을 확인하세요.  

```python
class Squirrel:
    species = "카피바라"  # 클래스 속성

    def __init__(self, name, current_job): #인스턴스 속성 magic method -> 이 객체가 가지고 있는 속성정의 생성자 -> 나만의 속성 정의
        self.name = name
        self.current_job = current_job

    def introduce(self): #객체가 가지고 있는 행동
        return f"저는 {self.name}이고, 현재 직업은 {self.current_job}입니다."


capy1 = Squirrel("형준", "학생") # 속성 세가지 species , name current job
capy2 = Squirrel("몽실", "개발자") # 속성 세가지 --

print(capy1.introduce())
print(capy2.introduce())

print(capy1.species)
print(capy2.species)
```

Character 클래스를 생성하세요.  
클래스 속성 skills를 정의하여 빈 리스트로 초기화하세요.  
생성자에서 인스턴스 속성 skills를 별도로 초기화하세요.  
add_skill 메서드에서 클래스 속성이 아닌 인스턴스 속성에 스킬을 추가하도록 수정하세요.  

```python
class Character:
    skills = []  # 클래스 속성

    def __init__(self):
        self.skills = []  # 인스턴스 속성 생성자에서 별도로 초기화 하기 빈값으로 두기

    def add_skill(self, skill): #클래스 속성이 아닌 인스턴스 속성에 스킬을 추가하도록 수정하기 객체=인스턴스
        self.skills.append(skill)


c1 = Character()
c2 = Character()

c1.add_skill("Python")
c2.add_skill("Java")

print(c1.skills)
print(c2.skills)
```

문제 4: 네임스페이스 테스트 - 전역과 로컬  
카피바라는 이직 준비 도구를 사용하며, 전역 변수와 로컬 변수를 관리합니다. namespace_test 함수를 작성해 네임스페이스를 테스트하세요.  
전역 변수 tool을 "연필"로 설정하세요.  
 함수 namespace_test 내부에 로컬 변수 local_tool을 "키보드"로 설정하고, global 키워드를 사용해 전역 변수를 "펜"으로 수정하세요.  
함수 내부에서 로컬 변수와 전역 변수의 값을 출력하세요.  

```python
tool = "연필"  # 전역 변수
print("함수 실행 전 전역 변수:" , tool)

def namespace_test():
   

    local_tool = "키보드"  # 지역 변수
    global tool
    tool = "펜"           # 전역 변수 수정

    print("로컬 변수:", local_tool)
    print("전역 변수:", tool)

namespace_test()

print("함수 실행 후 전역 변수:", tool)
```

문제 5: 네임스페이스 테스트 - 클래스와 인스턴스  
카피바라는 이직 준비 중 클래스를 통해 도구를 관리합니다. Toolbox 클래스를 만들어 네임스페이스를 테스트하세요.  
클래스 속성 tools를 ["노트북", "책상"]으로 초기화하세요.  
생성자를 통해 인스턴스 속성 tools를 설정하고, 객체마다 다른 도구를 가질 수 있도록 설정하세요.  
add_tool 메서드를 만들어 인스턴스 속성에 새 도구를 추가하세요.  

```python
class Toolbox:
    tools = ["노트북", "책상"]  # 클래스 속성

    def __init__(self):
        self.tools = []  # 인스턴스 속성

    def add_tool(self, tool):
        Toolbox.tools.append(tool)  # 클래스 속성에 추가


t1 = Toolbox()
t2 = Toolbox()

t1.add_tool("태블릿")
t2.add_tool("스마트폰")

print("클래스 속성:", Toolbox.tools)
print("t1 인스턴스:", t1.tools)
print("t2 인스턴스:", t2.tools)
```

카피바라는 이직 준비 중 추가적인 정보가 필요합니다. 중첩 함수와 nonlocal 키워드를 사용해 바깥 함수의 변수를 수정하세요.

바깥 함수의 변수 status를 "작성 중"으로 초기화하세요.  
중첩 함수에서 status를 "완료"로 수정하세요.  

```python
def nested_function() :
    status = "작성중"

    def inner_nested_function() :
        nonlocal status
        status = "완료"
    
    print(status)
    inner_nested_function()
    print(status)

nested_function()
```

---

1. `Person` 클래스:
    - 클래스 속성 `national`은 "Korea"로 설정합니다.
    - 메서드 `introduce`는 `"안녕하세요, 저는 한국에서 왔습니다."`를 반환합니다.
2. `Candidate` 클래스:
    - `Person` 클래스를 상속받습니다.
    - 생성자에서 이름(`name`)과 관심 분야(`field`)를 매개변수로 받아 저장합니다.

### 문제 2: **다중 상속을 활용한 클래스 생성**

카피바라는 이직을 준비하며 여러 강사님들의 도움을 받고 있습니다. 강사의 역할을 분리한 클래스를 만들어 상속 구조를 테스트하세요.

1. `Mentor` 클래스:
    - 메서드 `guide`는 `"이직 준비를 도와드립니다."`를 반환합니다.
2. `Lecturer` 클래스:
    - 메서드 `teach`는 `"이직에 필요한 기술을 가르쳐드립니다."`를 반환합니다.
3. `Coach` 클래스:
    - `Mentor`와 `Lecturer` 클래스를 다중 상속받습니다.
    - 메서드 `support`는 `"여러 방면으로 지원을 아끼지 않겠습니다."`를 반환합니다.

---

```python
class Mentor :
    def guide(self) :
        return "이직 준비를 도와드립니다."

class Lecturer : 
    def teach(self) :
        return "이직에 필요한 기술을 가르쳐드립니다."

class Coach(Mentor,Lecturer) :
    def support(self) :
        return "여러 방면으로 지원을 아끼지 않겠습니다."

coach = Coach()
print(coach.guide())
print(coach.teach())
print(coach.support())
```

### 문제 3: **부모 클래스 속성 오버라이딩**

카피바라는 이직 준비를 위해 개인적인 특성을 강조하고 있습니다. 부모 클래스의 메서드를 오버라이딩하여 각기 다른 인스턴스를 만들어보세요.

1. `Person` 클래스:
    - 속성 `name`과 `national`을 생성자로 받아 저장합니다.
    - 메서드 `introduce`는 `"안녕하세요, 저는 [name]입니다. 저는 [national]에서 왔습니다."`를 반환합니다.
2. `Candidate` 클래스:
    - `Person` 클래스를 상속받습니다.
    - `introduce` 메서드를 오버라이드하여 `"안녕하세요, 제 이름은 [name]이고, 이직 준비 중입니다."`를 반환합니다.

```python
class Person :
    def __init__(self, name, national) :
        self.name = name
        self.national = national

    def introduce(self) :
        return f"안녕하세요, 저는 {self.name}입니다. 저는 {self.national}에서 왔습니다."

class Candidate(Person) :
    def introduce(self) :
        super().introduce() #부모메서드 호출
        return f"안녕하세요, 제 이름은 {self.name}이고, 이직 준비 중입니다."

person = Person('임형준','한국')
candidate = Candidate('임형준','한국')

print(person.introduce())
print(candidate.introduce())
```

### 문제 4: **상속의 특성 통합 테스트**

카피바라는 강사님들의 지도를 받으며 다양한 분야에서 경험을 쌓고 있습니다. 상속과 메서드 오버라이딩, 다중 상속을 종합적으로 테스트하세요.

1. `Mentor` 클래스:
    - 속성 `field`를 생성자로 받아 저장합니다.
    - 메서드 `guide`는 `"저는 [field] 분야의 멘토입니다."`를 반환합니다.
2. `Lecturer` 클래스:
    - 속성 `experience`를 생성자로 받아 저장합니다.
    - 메서드 `teach`는 `"저는 [experience]년의 강의 경력을 가지고 있습니다."`를 반환합니다.
3. `Candidate` 클래스:
    - `Mentor`와 `Lecturer` 클래스를 다중 상속받습니다.
    - `guide` 메서드를 오버라이드하여 `"저는 [field] 분야의 강사이며, 이직 준비를 도와드립니다."`를 반환합니다.

```python
class Mentor :
    def __init__(self, field) :
        self.field = field

    def guide(self) :
        return f"저는 {self.field} 분야의 멘토입니다."

class Lecturer :
    def __init__(self, experience) :
        self.experience = experience

    def teach(self) :
        return f"저는 {self.experience}년의 강의 경력을 가지고 있습니다."

class Candidate(Mentor , Lecturer) :
    
    def teach(self) :
        super().teach()
        return f"저는 {self.experience}년의 강의 경력을 가지고 있습니다."
        
    def guide(self) :
        super().guide()
        return f"저는 {self.field} 분야의 강사이며, 이직 준비를 도와드립니다."

mentor = Mentor('AX')
lecturer = Lecturer(10)
candidate = Candidate('AX')

print(mentor.guide())
print(lecturer.teach())
print(candidate.guide())
```
