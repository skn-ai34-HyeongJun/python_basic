```python
def cal(a, b):
    return a+b, a-b

print(cal(2,3))
print(type(cal(2,3)))
```

```python
def add_many(*args):
    result = 0

    for i in args :
        result = result + i
    
    return result

print(add_many(1,2,3,4,5))
```

```python
def sum(a,b,c) :
    return a + b + c

numbers = [1,2,3]
#sum(numbers) a 하나만 전달

print(sum(*numbers)) # sum(1, 2, 3)
```

```python
numbers = [ 1, 2, 3, 4, 5]
squared_numbers = map(lambda x : x**2, numbers)
print(list(squared_numbers))
```

```python
numbers = [ 1, 2, 3, 4, 5, 6]
even_numbers = filter(lambda x : x%2 == 0, numbers) #람다 함수를 고차함수의 인자로 전달 (배열 요소 필터링)
print(list(even_numbers))
```

```python
points = [(1,2), (3,1), (5,-1)]
sorted_points = sorted(points, key=lambda x : x[1]) #배열 정렬 기준을 람다 함수로 전달
print(sorted_points) 
```

```python
students = [
    {"name":"bear", "score":90},
    {"name":"tiger", "score":70},
    {"name":"rabbit", "score":85}
]

# 점수 기준 정렬
print(sorted(students, key=lambda x: x["score"]))
```
